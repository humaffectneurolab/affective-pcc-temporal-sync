## Cell 1. Set paths and load helper

```python
from __future__ import annotations

import importlib.util
from pathlib import Path
import sys

# Run this notebook from the isrsa/ folder, or change CODE_DIR to that folder.
CODE_DIR = Path.cwd().resolve()
DATA_DIR = CODE_DIR.parents[1] / "data"
RESULTS_DIR = CODE_DIR.parents[1] / "results"

HELPER_PATH = CODE_DIR / "isrsa_functions.py"
spec = importlib.util.spec_from_file_location("isrsa_functions", HELPER_PATH)
if spec is None or spec.loader is None:
    raise ImportError(f"Cannot import {HELPER_PATH}")

ir = importlib.util.module_from_spec(spec)
sys.modules["isrsa_functions"] = ir
spec.loader.exec_module(ir)

ir.setup(data_dir=DATA_DIR, results_dir=RESULTS_DIR)

# Optional local exclusions (do not commit this file):
# ir.configure_exclusions(path="../local_config/exclusions.json")
```
